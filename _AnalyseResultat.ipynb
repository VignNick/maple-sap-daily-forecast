```python
years = [2017, 2024]
save_folder = "models_testing/best_models_graphs/"

# Charger le modèle et le data_module correspondant à cette expérience
from models.acer_conf_lstm import AcerConfLSTM
from models.TSMixer import TSMixerExt, TSMixerExt_PositiveOutput, TSMixerExt_Config, NormPlacement
from models.CustomMixer import MeteoTSMixer
from models.model import MODEL_REGISTRY
from models.acer_conf_lstm import AcerConfLSTM
from models.mlp_block import MLPBlock
import models.activations as act_f
import torch
from classes.ACER_data_handler import ACERDataModule, ACERDataset
import utils.SapGraph as sapgraph
from IPython.display import display
import os
import pandas as pd

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
dtype_use = torch.bfloat16

just_load_metrics = True
all_metrics = {}

for graph in graph_to_generate:
    # Charger le modèle
    ckpt = torch.load(graph['path'], weights_only=False)
    
    model_config = ckpt["model_config"]
    loaded_model = MODEL_REGISTRY[ckpt["model_class"]].from_dict(ckpt["model_config"])
    loaded_model.load_state_dict(ckpt["model_state_dict"])
    loaded_model.to(device, dtype_use)
    loaded_model.eval()

    loaded_data_module = ACERDataModule.from_dict(ckpt["data_module_config"])
    loaded_data_module.get_ready()
    loaded_validation_dataset = loaded_data_module.gen_val_dataset()

    # Les métriques de performance ---------------------------------------------------------
    loaded_validation_dataset.ensure_precached(device, dtype_use)
    _, _, metrics = sapgraph.__get_prediction_and_real_values(loaded_model, loaded_validation_dataset)
    metrics_text = (
            f"{graph['Model']} - Horizon {graph['Horizon']} jours\n"
            "---------------------------------------------------\n"
            f"MSE:     {metrics['MSE']:.4f}\n"
            f"RMSE:    {metrics['RMSE']:.4f}\n"
            f"RMSE%:   {metrics['RMSE%']:.4f}%\n"
            f"WAPE:    {metrics['WAPE']:.4f}%\n"
            f"R^2:     {metrics['R2']:.4f}"
        )
    # print(f"Metrics for {graph['Model']} - Horizon {graph['Horizon']} jours:\n{metrics_text}")
    all_metrics[f"{graph['Model']}_horizon_{graph['Horizon']}"] = metrics

    # Sauvegarder les métriques dans un fichier texte
    metrics_save_path = os.path.join(save_folder, f"{graph['Model']}/{graph['Model']}_horizon_{graph['Horizon']}_metrics.txt")
    os.makedirs(os.path.dirname(metrics_save_path), exist_ok=True)
    with open(metrics_save_path, 'w') as f:
        f.write(metrics_text)

    if just_load_metrics:
        continue  # Passer à l'itération suivante si on ne veut que charger les métriques

    # Générer le graphique contenant toutes les années spécifiées ---------------------------
    acerDataSet = loaded_data_module.get_first_system_for_years(years)
    acerDataSet.ensure_precached(device, dtype_use)

    fig, ax = sapgraph.sap_graph(
        model= loaded_model,
        dataset=acerDataSet,
        horizon=graph["Horizon"],
        figsize=(16, 6),
        dpi=500,
        showDates=7,
        datesAngle=60,
        sup_text=graph["sup_text"],
        show_metrics=True,
        metrics=metrics,
        title=f"Courbe des valeurs réelles et prédites\n{graph['Model']} - Horizon {graph['Horizon']} jours pour les années {years}"
        )

    fig.tight_layout()  # Ajuster la mise en page pour éviter le chevauchement des éléments
    ax.autoscale(enable=True, axis='x', tight=True)
    # Sauvegarder la figure dans le dossier de l'expérience
    save_path = os.path.join(save_folder, f"{graph['Model']}/{graph['Model']}_horizon_{graph['Horizon']}_validation.png")
    os.makedirs(os.path.dirname(save_path), exist_ok=True)
    fig.savefig(save_path, bbox_inches='tight')
    print(f"{graph['Model']}_horizon_{graph['Horizon']}_years_{years} : Graph saved to {save_path}")
    # display(fig)

    # Générer un graphique pour chaque année spécifiée ------------------------------------------------
    for year in years:
        acerDataSet = loaded_data_module.get_first_system_of_year_dataset(year)
        acerDataSet.ensure_precached(device, dtype_use)

        fig, ax = sapgraph.sap_graph(
            model=loaded_model,
            dataset=acerDataSet,
            horizon=graph["Horizon"],
            figsize=(10, 6),
            dpi=500,
            showDates=7,
            datesAngle=15,
            sup_text=None,
            show_metrics=False,
            # title=f"Courbe des valeurs réelles et prédites\n{graph['Model']} - Horizon {graph['Horizon']} jours pour {year}"
            title=f"{graph['Model']} - Horizon {graph['Horizon']} jours pour {year}"
            )

        fig.tight_layout()  # Ajuster la mise en page pour éviter le chevauchement des éléments
        ax.autoscale(enable=True, axis='x', tight=True)

        # Sauvegarder la figure dans le dossier de l'expérience
        save_path = os.path.join(save_folder, f"{graph['Model']}/{graph['Model']}_horizon_{graph['Horizon']}_year_{year}.png")
        os.makedirs(os.path.dirname(save_path), exist_ok=True)
        fig.savefig(save_path, bbox_inches='tight')
        print(f"{graph['Model']}_horizon_{graph['Horizon']}_year_{year} : Graph saved to {save_path}")
        # display(fig)
```

```python
import pandas as pd
from IPython.display import display

df_all_metrics = pd.DataFrame(all_metrics).T

# Extraire l'horizon depuis l'index
df_all_metrics['Horizon'] = (
    df_all_metrics.index
    .str.extract(r'horizon_(\d+)')[0]
    .astype(int)
    .to_numpy()
)

# Métriques où PLUS PETIT = meilleur
min_metrics = ['MSE', 'RMSE', 'RMSE%', 'WAPE']

# Métriques où PLUS GRAND = meilleur
max_metrics = ['R2']


def highlight_ranking(df):
    styles = pd.DataFrame('', index=df.index, columns=df.columns)
    # Color gradient de vert à bleu à jaune à rouge pour les métriques à minimiser
    gradient_colors = ['green', 'lightgreen', 'lightblue', 'khaki', 'gold', 'lightcoral', 'red']

    sort_types = [{'ascending': True, 'metrics': min_metrics}, {'ascending': False, 'metrics': max_metrics}]

    for sort_type in sort_types:
        for col in sort_type['metrics']:
            ranked = df[col].sort_values(ascending=sort_type['ascending'])
            
            for i in range(min(len(ranked), len(gradient_colors))//2):
                styles.loc[ranked.index[i], col] = (
                    f'background-color: {gradient_colors[i]}; font-weight: bold'
                )
    
                styles.loc[ranked.index[-i-1], col] = (
                    f'background-color: {gradient_colors[-i-1]}; font-weight: bold'
                )
    
            if len(gradient_colors)%2 and len(ranked)%2 :
                styles.loc[ranked.index[len(ranked)//2], col] = (
                    f'background-color: {gradient_colors[len(gradient_colors)//2]}; font-weight: bold'
                )

    return styles


# Affichage par horizon
for horizon, df in df_all_metrics.groupby('Horizon'):

    print(f"\n{'='*60}")
    print(f"HORIZON {horizon}")
    print(f"{'='*60}")

    display(
        df.style
        .apply(highlight_ranking, axis=None)
        .format({
            'MSE': '{:.4f}',
            'RMSE': '{:.4f}',
            'RMSE%': '{:.4f}',
            'WAPE': '{:.4f}',
            'R2': '{:.4f}'
        })
    )
```
